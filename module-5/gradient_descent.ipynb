```python
import numpy as np
import matplotlib.pyplot as plt

# Original func
def f(x):
    return x**2

#  Deriative
def dx(x):
    return 2 * x

x_start = 4
learning_rate = 0.1
# learning_rate = 1
# learning_rate = 0.01
# learning_rate = 1.05



epohs = 20

x_history = []
y_history = []

x_current = x_start

for _ in range(epohs):
    y = f(x_current)
    x_history.append(x_current)
    y_history.append(y)
    
    #  Gradient
    gradient = dx(x_current)
    x_current = x_current - learning_rate * gradient
    
    
x_values = np.linspace(-6, 6, 100)
y_values = f(x_values)

plt.plot(x_values, y_values, label='f(x) = x^2')
plt.scatter(x_history, y_history, label='Кроки градієнтного спуску', color='red')
plt.plot(x_history, y_history, color='red', linestyle='--')
plt.title('Градієнтний спуск для квадратичної функції')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
    
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. СТВОРЕННЯ "ЯМИ" (ВІЗУАЛІЗАЦІЯ РЕЛЬЄФУ)
# Робимо осі x та y
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

# Створюємо 2D-сітку координат (ось та сама магія meshgrid)
X, Y = np.meshgrid(x, y)

# Наша функція помилки (скалярна функція векторного аргументу)
Z = X**2 + Y**2

# Налаштовуємо 3D-графік
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d') # Кажемо matplotlib, що хочемо 3D

# Малюємо поверхню (cmap='viridis' просто дає красивий градієнт кольорів)
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.6, edgecolor='none')

# ==========================================
# ТУТ МАЄ БУТИ ТВІЙ КОД ГРАДІЄНТНОГО СПУСКУ
# ==========================================
x_history = []
y_history = []
z_history = []
x_0 = 4
y_0 = 4
epochs = 20

x_curr = x_0
y_curr = y_0

def z(x,y):
    return x**2 + y**2

def df_dx(x):
    return 2*x

def df_dy(y):
    return 2*y

lr = 0.1

for _ in range(epochs):
    x_history.append(x_curr)
    y_history.append(y_curr)
    z_value = z(x_curr, y_curr)
    z_history.append(z_value)
    
    x_curr = x_curr - lr * df_dx(x_curr)
    y_curr = y_curr - lr * df_dy(y_curr)
    
   
   

ax.plot(x_history, y_history, z_history, color='red', marker='o', linewidth=2)

ax.set_title("3D Градієнтний спуск")
ax.set_xlabel("Вісь X (Вага 1)")
ax.set_ylabel("Вісь Y (Вага 2)")
ax.set_zlabel("Вісь Z (Помилка / Loss)")

plt.show()
```
