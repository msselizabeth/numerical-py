#Теорема Баєса

**Приклад.**
Нехай ми маємо тест на деяке захворювання, який має чутливість 99% (імовірність позитивного результату, коли захворювання є). Про захворювання відомо, що воно зустрічається в однієї людини з 10000, тобто ймовірність захворювання — 0.0001.

Питання: Яка ймовірність, що особа справді хвора, якщо тест показав позитивний результат?

Нехай подія $A$ — пацієнт хворий. Подія  $B$ — отримано позитивний тест. Розглянемо наступні ймовірності.

1. $ P(A) $ — апріорна ймовірність бути хворим, тобто $ P(\text{Хворий}) = 0.0001 $ (апріорна ймовірність наявності захворювання).
2. $ P(\neg A) = 0.9999$ — апріорна ймовірність не хворіти на цю хворобу.
2. $ P(B|A) $ — імовірність отримати позитивний результат при хворобі, тобто $ P(\text{Позитивний тест}|\text{Хворий}) = 0.99 $.
3. $ P(B|\neg A) $ — імовірність отримати позитивний результат при відсутності хвороби, тобто $ P(\text{Позитивний тест}|\text{Відсутність хвороби}) = 1 - 0.99 = 0.01 $.

Загальна ймовірність позитивного тесту в даних не наводиться, однак її можна обчислити, користуючись формулою повної ймовірності:

$P(B)=P(B|A)\cdot P(A) + P(B|\neg A)\cdot P(\neg A) = (0.99 \cdot 0.0001)+(0.01 \cdot 0.9999) = 0.0101$,

отже, маємо розширену формулу Баєса:

$\displaystyle P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B|A)\cdot P(A) + P(B|\neg A)\cdot P(\neg A)}. $

Тепер можна використовувати формулу Баєса для обчислення $ P(\text{Хворий}|\text{Позитивний тест}) $:

$\displaystyle P(\text{Хворий}|\text{Позитивний тест}) = \frac{P(\text{Позитивний тест}|\text{Хворий}) \cdot P(\text{Хворий})}{P(\text{Позитивний тест})} $

$\displaystyle P(\text{Хворий}|\text{Позитивний тест}) = \frac{0.99 \cdot 0.0001}{0.0101} = 0.0098 $

Отже, апостеріорна ймовірність того, що особа справді хвора при позитивному тесті, складає менше 1%.

```python
def bayes_theorem(p_a, p_b_given_a, p_b_given_not_a):
    """
    Обчислює апостеріорну ймовірність за допомогою теореми Баєса.

    :param p_a: Початкова ймовірність A (наприклад, наявність хвороби)
    :param p_b_given_a: Ймовірність B, якщо A (наприклад, позитивний тест при наявності хвороби)
    :param p_b_given_not_a: Ймовірність B, якщо не A (наприклад, позитивний тест при відсутності хвороби)
    :return: Обчислена ймовірність A, якщо B (наприклад, ймовірність хвороби при позитивному тесті)
    """
    p_not_a = 1 - p_a
    p_b = (p_b_given_a * p_a) + (p_b_given_not_a * p_not_a)

    if p_b == 0:
        raise ValueError("Помилка: ймовірність P(B) не може бути 0.")

    return (p_b_given_a * p_a) / p_b

# Приклад використання
p_disease = 0.0001
p_positive_given_disease = 0.99
p_positive_given_no_disease = 0.01

# Розрахунок імовірності захворювання при позитивному тесті
p_disease_given_positive = bayes_theorem(p_disease, p_positive_given_disease, p_positive_given_no_disease)
print(f"Ймовірність захворювання при позитивному тесті: {p_disease_given_positive:.6f}")
```

#Класифікатор спама

Визначимо деякі навчальні та тестові дані для кожного класу: спаму та важливих повідомлень. Ці повідомлення будуть короткими для стислості та ясності.

> 🟡
> ЗМЕНШІТЬ СЛОВНИКИ ДЛЯ НАГЛЯДНОСТІ

```python
# 📌 Тренувальні спам-листи
train_spam = [
    "send us your password", "review our website", "send your password",
    "send us your account", "limited time offer", "win a free iPhone",
    "click here to claim your reward", "your loan is pre-approved",
    "urgent response required", "you have won the lottery",
    "dear customer, your account has been suspended",
    "100% free trial", "this is not a scam", "earn money fast",
    "get rich quick scheme", "exclusive deal for you",
    "cheap medications online", "instant cash now"
]

# 📌 Тренувальні ham-листи
train_ham = [
    "your activity report", "benefits physical activity", "the importance vows",
    "let’s schedule a meeting", "your bank statement is ready",
    "project update from the team", "please review the latest changes",
    "can we reschedule our call?", "here is the document you requested",
    "happy birthday, wishing you all the best!", "see you at the conference",
    "weekly team report", "your travel itinerary", "thank you for your purchase",
    "your recent transaction details", "important updates for employees",
    "family gathering next weekend", "doctor appointment confirmation"
]

# 📌 Тестові email-и
test_emails = {
    'spam': [
        "renew your password", "renew your vows", "win a free laptop",
        "click now for a special offer", "congratulations, you have won a prize",
        "instant loan approval", "your account has been locked",
        "get paid for taking surveys", "claim your free vacation",
        "free cash waiting for you"
    ],
    'ham': [
        "benefits of our account", "the importance of physical activity",
        "can you send me the report?", "meeting rescheduled for tomorrow",
        "weekly newsletter subscription", "thank you for your support",
        "invoice for your recent order", "your payment has been processed",
        "family dinner this weekend", "don’t forget about the project deadline"
    ]
}

# 📊 Вивід розміру тренувальних та тестових наборів
print(f"📩 Кількість тренувальних spam-листів: {len(train_spam)}")
print(f"✅ Кількість тренувальних ham-листів: {len(train_ham)}")
print(f"🔍 Кількість тестових spam-листів: {len(test_emails['spam'])}")
print(f"🔎 Кількість тестових ham-листів: {len(test_emails['ham'])}")
```

1. Проводимо ітерацію за розміченими спам-повідомленнями і для кожного слова $w$ у всьому навчальному наборі підраховуємо, скільки спам-повідомлень містить $w$.

2. Перетворюємо кожен елемент списку у множину для виділення унікальних слів — та назад у список.

```python
# Отримуємо унікальні слова, що зустрічаються в спам-листах
vocab_words_spam = [word.lower() for sentence in train_spam for word in sentence.split()]
vocab_unique_words_spam = list(set(vocab_words_spam))

print(vocab_words_spam, '\n\n')
print(f"Унікальні слова в спамі: {vocab_unique_words_spam}")
```

Визначаємо ймовірність приналежності кожного слова до спаму та застосовуємо згладжування.

```python
dict_spamicity = {}
total_spam = len(train_spam)

for w in vocab_unique_words_spam:
    emails_with_w = sum(1 for sentence in train_spam if w in sentence.lower())  # Лічильник входжень слова у спам
    spamicity = (emails_with_w + 1) / (total_spam + 2)  # Laplace smoothing

    print(f"Слово '{w}' зустрічається в {emails_with_w} спам-листах.")
    print(f"Spamicity для '{w}': {spamicity:.4f}\n")

    dict_spamicity[w] = spamicity

print(f"📊 Словник spamicity: {dict_spamicity}")
```

Визначаємо ймовірність приналежності кожного слова до важливих повідомлень.

```python
# make a vocabulary of unique words that occur in known ham emails
vocab_words_ham = [word.lower() for sentence in train_ham for word in sentence.split()]
vocab_unique_words_ham = list(set(vocab_words_ham))

print(f"Унікальні слова в ham-листах: {vocab_unique_words_ham}")

```

```python
dict_hamicity = {}
total_ham = len(train_ham)

for w in vocab_unique_words_ham:
    emails_with_w = sum(1 for sentence in train_ham if w in sentence.lower())  # Підрахунок входжень слова в ham
    hamicity = (emails_with_w + 1) / (total_ham + 2)  # Laplace smoothing

    print(f"Слово '{w}' зустрічається в {emails_with_w} ham-листах.")
    print(f"Hamicity для '{w}': {hamicity:.4f}\n")

    dict_hamicity[w] = hamicity

print(f"📊 Словник hamicity: {dict_hamicity}")

```

Обчислюємо ймовірність спаму $P(S)$.

```python
prob_spam = len(train_spam) / (len(train_spam)+(len(train_ham)))
print(prob_spam)
```

Обчислюємо ймовірність важливого повідомлення  $P(H)$.

```python
prob_ham = len(train_ham) / (len(train_spam)+(len(train_ham)))
print(prob_ham)
```

```python
prob_ham  + prob_spam
```

Для повідомлень без розмітки (тестові повідомлення) створюємо список слів.

```python
# Збір тестових email-ів (об'єднання списків spam і ham)
tests = []
tests.extend(test_emails['spam'])
tests.extend(test_emails['ham'])

print(f"📩 Тестові email-и: {tests}")

# Розбиваємо кожен email на список окремих слів (токенізація)
distinct_words_as_sentences_test = [sentence.split() for sentence in tests]

print(f"\n🔤 Тестові email-и у вигляді списків слів:")
for i, email in enumerate(distinct_words_as_sentences_test):
    print(f"{i+1}. {email}")

# Поділ на тестові групи (Spam / Ham)
test_spam_tokenized = distinct_words_as_sentences_test[:len(test_emails['spam'])]
test_ham_tokenized = distinct_words_as_sentences_test[len(test_emails['spam']):]

print(f"\n🛑 Токенізовані тестові spam email-и: {test_spam_tokenized}")
print(f"✅ Токенізовані тестові ham email-и: {test_ham_tokenized}")
```

Видаляємо слова, що не зустрічались у навчальному наборі даних.

```python
def filter_test_sentences(test_tokenized, vocab_spam, vocab_ham, label):
    """
    Фільтрує тестові речення, залишаючи лише слова, що є у списках spam або ham.

    :param test_tokenized: список токенізованих тестових листів
    :param vocab_spam: унікальні слова, що зустрічаються у спам-листах
    :param vocab_ham: унікальні слова, що зустрічаються у ham-листах
    :param label: тип тесту ("spam" або "ham")
    :return: відфільтрований список речень
    """
    reduced_sentences = []

    for sentence in test_tokenized:
        filtered_words = [word for word in sentence if word in vocab_spam or word in vocab_ham]

        # Вивід процесу фільтрації
        print(f"\n🔍 Тестовий {label} email: {sentence}")
        for word in sentence:
            if word in vocab_spam or word in vocab_ham:
                print(f"✅ '{word}', залишаємо")
            else:
                print(f"❌ '{word}', слово не зустрічається в тренувальних spam/ham даних")

        reduced_sentences.append(filtered_words)

    return reduced_sentences


# 📌 Фільтрація spam і ham тестових листів
reduced_sentences_spam_test = filter_test_sentences(test_spam_tokenized, vocab_unique_words_spam, vocab_unique_words_ham, "spam")
reduced_sentences_ham_test = filter_test_sentences(test_ham_tokenized, vocab_unique_words_ham, vocab_unique_words_spam, "ham")

```

Вилучення неключових слів. Це може допомогти класифікатору зосередитись на важливих словах.

```python
def remove_non_key_words(filtered_sentences, non_key_words, label):
    """
    Видаляє непотрібні ("non-key") слова зі списку тестових речень.

    :param filtered_sentences: список тестових листів після фільтрації
    :param non_key_words: список стоп-слів для видалення
    :param label: тип тесту ("spam" або "ham")
    :return: очищений список тестових листів
    """
    stemmed_sentences = [
        [word for word in sentence if word not in non_key_words]
        for sentence in filtered_sentences
    ]

    # Вивід процесу очищення
    print(f"\n🧹 Очищення {label} email-ів:")
    for original, cleaned in zip(filtered_sentences, stemmed_sentences):
        removed_words = set(original) - set(cleaned)
        print(f"📩 {original} ➝ {cleaned} (Видалено: {removed_words})")

    return stemmed_sentences


# 📌 Видалення "non-key" слів (стоп-слів)
non_key_words = ['us', 'the', 'of', 'your']
test_spam_stemmed = remove_non_key_words(reduced_sentences_spam_test, non_key_words, "spam")
test_ham_stemmed = remove_non_key_words(reduced_sentences_ham_test, non_key_words, "ham")

# 📊 Вивід кінцевого результату
print(f"\n🚀 **Фінальні токенізовані spam email-и:** {test_spam_stemmed}")
print(f"✅ **Фінальні токенізовані ham email-и:** {test_ham_stemmed}")
```

Розглянемо приклад реалізації наївного класифікатора Баєса. Його «наївність» полягає в тому, що він не враховує зв’язок одного слова з іншими. Він припускає, що слова, які з’являються в реченні, незалежні одне від одного. Це, звичайно, не так, тому ця рання форма фільтрації спаму була витіснена більш просунутими формами аналізу текстів, такими, які використовують векторизацію слів.

```python
from math import log, exp

def mult_log(probs):
    """Обчислює логарифм добутку ймовірностей."""
    return sum([log(p) for p in probs if p > 0])

def Bayes(email, prob_spam, prob_ham, dict_spamicity, dict_hamicity, total_spam, total_ham):
    probs_s = []
    probs_h = []

    for word in email:
        try:
            pr_WS = dict_spamicity[word]  # Ймовірність слова в спамі
        except KeyError:
            pr_WS = 1 / max(1, total_spam + 2)  # Laplace smoothing

        try:
            pr_WH = dict_hamicity[word]  # Ймовірність слова в нормальних листах
        except KeyError:
            pr_WH = 1 / max(1, total_ham + 2)  # Laplace smoothing

        probs_s.append(pr_WS)
        probs_h.append(pr_WH)

    # Використовуємо логарифми замість прямого множення
    log_P_S = log(prob_spam) + mult_log(probs_s)
    log_P_H = log(prob_ham) + mult_log(probs_h)

    # Обчислюємо фінальну ймовірність спаму
    final_classification = exp(log_P_S) / (exp(log_P_S) + exp(log_P_H))

    print('###########################')
    if final_classification >= 0.5:
        print(f'📩 Email класифіковано як **SPAM** з довірою {final_classification * 100:.2f}%')
    else:
        print(f'📩 Email класифіковано як **HAM** з довірою {final_classification * 100:.2f}%')

    return final_classification
```

```python
results_spam = []

# Тестування кожного повідомлення
for email in test_spam_stemmed:
    print("\n")
    print(f"📩 Testing stemmed SPAM email: {email}")
    spam_confidence = Bayes(email, prob_spam, prob_ham, dict_spamicity, dict_hamicity, total_spam, total_ham)
    print(f"🔥 Spam confidence: {spam_confidence:.4f}")

    # Збереження у список
    results_spam.append((email, spam_confidence))

# Вивід усіх результатів
print("\n📊 **Summary of Tests**")
for email, conf in results_spam:
    classification = "SPAM" if conf >= 0.5 else "HAM"
    print(f"📩 {email} ➝ {classification} ({conf*100:.2f}%)")
```

```python
# Збереження результатів
results_ham = []

# Тестування кожного повідомлення
for email in test_ham_stemmed:
    print("\n")
    print(f"📩 Testing stemmed HAM email: {email}")
    spam_confidence = Bayes(email, prob_spam, prob_ham, dict_spamicity, dict_hamicity, total_spam, total_ham)
    print(f"🔥 Spam confidence: {spam_confidence:.4f}")

    # Збереження у список
    results_ham.append((email, spam_confidence))

# Вивід усіх результатів
print("\n📊 **Summary of HAM Tests**")
for email, conf in results_ham:
    classification = "HAM" if conf < 0.5 else "SPAM"
    print(f"📩 {email} ➝ {classification} ({conf*100:.2f}%)")
```
